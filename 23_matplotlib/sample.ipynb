```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("loan_approval_dataset_200_rows.csv")
```

```python
df
```

```python
df["employment_type"].value_counts().plot(kind="pie", color="steelblue", autopct = '%0.2f%%')

plt.title("Applicants by Employment Type")
plt.xlabel("Employment Type")
plt.ylabel("Applicant Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
