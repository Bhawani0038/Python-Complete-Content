```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
df
```

```python
df.head()
```

```python
sns.barplot(df, x = "sex", y = "total_bill", errorbar = None, estimator= "mean", hue = "time")
plt.show()
```

```python
sns.scatterplot(df, x = "total_bill", y = "tip", hue = 'sex', size = 'time', style = 'day')
```

```python
sns.histplot(df, x = "total_bill")
```

```python
sns.boxplot(df, y = 'tip', hue = 'time')
```

```python
sns.countplot(df, x = "sex", hue = 'time')
```
